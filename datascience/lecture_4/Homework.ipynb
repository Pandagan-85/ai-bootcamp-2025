# Exercises

1) Write a query to find first name and last name of employees whose first name:
- has 3 letters
- ends in a


```python
%%sql
SELECT first_name, last_name
FROM employees
WHERE first_name LIKE'__a'
```

2) Write a query to find, for each department:
- the name of the department
- the number of employees
- the number of employees who received a bonus
- the total amount of bonuses disbursed

```python
%%sql
SELECT d.name AS department, COUNT(e.id) as n_employees, COUNT(e.bonus) as n_employees_bonus, SUM(e.bonus) as total_bonus
FROM employees e
INNER JOIN departments d ON e.department_id = d.id
GROUP BY department
ORDER BY d.id
```

## Versione A
Dove non tocco i valori `null` e lo includo nel count

Uso `COUNT(*)` dando per scontato che i NULL equivalgano a 0

```python
%%sql
SELECT COUNT(*) as bonus_count, bonus

from employees
GROUP BY bonus
HAVING bonus_count > 1
```

Provo ad espandere la query con una `self join` per provare a recuperare nomi e id

```python
%%sql
SELECT 
    bonus,
    COUNT(*) AS bonus_count,
    GROUP_CONCAT(first_name) AS names,
    GROUP_CONCAT(id) AS employee_id
FROM employees
GROUP BY bonus
HAVING bonus_count > 1;
```

## Versione B
Converto i valori `null` in 0 e rifaccio la stessa cosa

```python
%%sql
SELECT 
    IFNULL(bonus,0) AS bonus_mod,
    COUNT(*) AS bonus_count,
    GROUP_CONCAT(first_name) AS names,
    GROUP_CONCAT(id) AS employee_id
FROM employees
GROUP BY bonus_mod
HAVING bonus_count > 1;
```

## Versione C
Converto i valori null in "N/A" e rifaccio la stessa cosa

```python
%%sql
SELECT 
    IFNULL(bonus,"N/A") AS bonus_mod,
    COUNT(*) AS bonus_count,
    GROUP_CONCAT(first_name) AS names,
    GROUP_CONCAT(id) AS employee_id
FROM employees
GROUP BY bonus_mod
HAVING bonus_count > 1;
```
